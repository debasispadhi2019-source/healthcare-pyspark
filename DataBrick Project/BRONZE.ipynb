```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
from pyspark.sql.types import *
import os
import pandas as pd 
import numpy as np
import datetime as dt
from pyspark.sql import SparkSession
```

## Bronze path

```python
BASE_PATH= "abfss://data@rgemrhealthcare.dfs.core.windows.net/"

BRONZE_PATH = BASE_PATH + "bronze/"

print(BRONZE_PATH)
```

# Read All CSV FIles

```python
##------DepartMent File ----------------##
department_schema ="""
            DeptID string,
            Name string,
            Location string,
            Specialization string
"""
##---------------PRovider ------------------------##
provider_schema="""
            ProviderID string,
            ProviderName string,
            Specialization string,
            DeptID string,
            Phone string,
            Email string,
            HireDate string
"""
##------------------Patient-----------------------##
patient_schema="""
            PatientID string,
            PatientName string,
            Gender string,
            DateOfBirth string,
            Phone string,
            Email string,
            Address string,
            BloodGroup string,
            InsuranceType string,
            RegistrationDate string"""

##--------------Encounter-------------------------##
encounter_schema= """
            EncounterID string,
            PatientID string,
            ProviderID string,
            EncounterDate string,
            EncounterType string,
            Status string,
            Diagnosis string,
            TreatmentAdvice string,
            FollowUpDate string"""

transaction_schema = """
    TransactionID string,
    EncounterID string,
    TransactionDate date,
    TransactionType string,
    Amount double,
    PaymentMethod string
"""


```

```python
departments = (
    spark.read.format('csv')\
        .option('header','true')
        .schema(department_schema)
        .load(f"{BRONZE_PATH}departments.csv")
)

providers = (
    spark.read.format('csv')\
        .option('header','true')
        .schema(provider_schema)
        .load(f"{BRONZE_PATH}providers.csv")
)

patients = (
    spark.read.format('csv')\
        .option('header','true')
        .schema(patient_schema)
        .load(f"{BRONZE_PATH}patients.csv")
)

encounters = (
    spark.read.format('csv')\
        .option('header','true')
        .schema(encounter_schema)
        .load(f"{BRONZE_PATH}encounters.csv")
)

transactions = (
    spark.read.format('csv')\
        .option('header','true')
        .schema(transaction_schema)
        .load(f"{BRONZE_PATH}transactions.csv")
)
```

## Bronze validation

```python
print("Departments :", departments.count())
print("Providers   :", providers.count())
print("Patients    :", patients.count())
print("Encounters  :", encounters.count())
print("Transactions:", transactions.count())
```
