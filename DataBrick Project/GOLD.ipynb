```python

BASE_PATH = "abfss://data@rgemrhealthcare.dfs.core.windows.net/"

SILVER_PATH= BASE_PATH + "silver/"
GOLD_PATH = BASE_PATH + "gold/"

print(SILVER_PATH)
```

## Read Silver Delta datasets

```python
encounter_enriched_df =(
    spark.read\
        .format('delta')
        .load(SILVER_PATH+'encounter_enriched')
)
#display('-----encounter_enriched_df-----')
display(encounter_enriched_df.limit(10))


provider_department_df = (
    spark.read
    .format("delta")
    .load(SILVER_PATH + "provider_dept")
)

display(provider_department_df.limit(10))


patient_encounter_df = (
    spark.read
    .format("delta")
    .load(SILVER_PATH + "encounter_patients")
)
display(patient_encounter_df.limit(10))

transaction_encounter_df = (
    spark.read
    .format("delta")
    .load(SILVER_PATH + "transaction_encounter")
)

display(transaction_encounter_df.limit(10))

```

## Verify the Silver data


```python
print("Encounter Enriched Count Records->", encounter_enriched_df.count())
print("Provider Department Count Records->", provider_department_df.count())
print("Patient Encounter Count Records->", patient_encounter_df.count())
print("Transaction Encounter Count Records->", transaction_encounter_df.count())
```
