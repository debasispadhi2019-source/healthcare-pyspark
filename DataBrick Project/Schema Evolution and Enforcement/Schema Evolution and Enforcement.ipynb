```python
from pyspark.sql.functions import *
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('Evo&enfoce').getOrCreate()

```

```python
df1=spark.createDataFrame(
    [
        (101, "Asha", "2026-09-01"),
        (102, "Ravi", "2026-09-02")
    ],
    ["patient_id", "patient_name", "admit_date"]
).withColumn("admit_date", to_date("admit_date"))

df1.display()

```

```python
%sql
create table if not exists shema_evo_enfor
```

```python
table_name='shema_evo_enfor'
df1.write.mode('overwrite').option('overwriteSchema', 'true').format('delta').saveAsTable(table_name)
print("Initial schema")
spark.read.table(table_name).printSchema()
spark.read.table(table_name).show()
```

## Sehema Enforcement 

```python
df2=spark.createDataFrame(
    [
        (101, "Asha", "2026-09-01",'xxxye'),
        (102, "Ravi", "2026-09-02",'yyyxxe')
    ],
    ["patient_id", "patient_name", "admit_date",'Insu_Pro']
).withColumn("admit_date", to_date("admit_date"))

df2.display()
```

```python
df2.write.mode('append').format('delta').saveAsTable(table_name)
print("Initial schema")
spark.read.table(table_name).printSchema()
spark.read.table(table_name).show()
```

## Schema Evolution

```python
df2.write.mode('append').option('mergeSchema','true').format('delta').saveAsTable(table_name)
print("Initial schema")
spark.read.table(table_name).printSchema()
spark.read.table(table_name).show()
```
